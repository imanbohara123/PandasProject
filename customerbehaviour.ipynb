```python
print(f'Maximum Annual Income = {df["annual_income"].max()}')
print(f'Minimum annual income = {df['annual_income'].min()}')
```

salary maximum minimum finding from the dataset

```python
df['income_segment'] = pd.qcut(
    df['annual_income'],
    q=3,
    labels=['low income', 'medium income', 'high income']
)
print(df[['income_segment','annual_income']])
```

Converting the annual_income into equal 4 segment

```python
df['spending_segment'] = pd.qcut(
    df['spending_score'],
    q=3,
    labels=['low spender', 'medium spender', 'high spender']
)
print(df[['spending_segment','spending_score']])
```

```python
df['age_segment'] = pd.cut(
    df['age'],
    bins=[0, 30, 50, 100],
    labels=['young', 'adult', 'senior'])
print(df[['age_segment','age']])
```
